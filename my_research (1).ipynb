## Графики

```python
grouped_data = full_train.groupby('experience')['price'].mean().reset_index()

# Строим столбчатую диаграмму
plt.figure(figsize=(10, 6))
plt.bar(grouped_data['experience'], grouped_data['price'])
plt.xlabel('Experience')
plt.ylabel('Average Price')
plt.title('Средняя цена по опыту')
plt.xticks(rotation=45)  # Поворот подписей оси X для удобства
plt.show()
```

### видим, что действительно платят за доп годы работы 

```python
# Список бинарных колонок
binary_columns = ['is_Agile', 'is_Android', 'is_Backend', 'is_Cloud',
                 'is_Code Review', 'is_Content/Copy', 'is_Data Science/ML',
                 'is_Databases', 'is_DevOps/SRE', 'is_DevRel',
                 'is_Entrepreneurship', 'is_Frontend', 'is_HR', 'is_Marketing',
                 'is_Product Management', 'is_Project Management', 'is_QA',
                 'is_System Design', 'is_Team Lead/Management', 'is_UX/UI/Design',
                 'is_iOS', 'is_Аналитика', 'is_Безопасность', 'is_Другое',
                 'is_Карьера', 'is_Сети', 'is_Собеседования']

# Вычисляем среднюю цену для каждой бинарной колонки
mean_prices = []
for col in binary_columns:
    mean_price = full_train.groupby(col)['price'].mean()
    # Берем среднее значение для случая, когда колонка = 1 (True)
    if 1 in mean_price.index:
        mean_prices.append(mean_price.loc[1])
    else:
        mean_prices.append(0)

# Создаем DataFrame для удобства визуализации
result_df = pd.DataFrame({
    'feature': binary_columns,
    'mean_price': mean_prices
})

# Сортируем по средней цене для лучшей визуализации
result_df = result_df.sort_values('mean_price', ascending=False)

# Строим график
plt.figure(figsize=(15, 10))
plt.barh(result_df['feature'], result_df['mean_price'])
plt.xlabel('Средняя цена')
plt.ylabel('Признак')
plt.title('Средняя цена по бинарным признакам')
plt.tight_layout()
plt.show()

##
```

## predictions

```python
from autogluon.tabular import TabularPredictor
label = 'price'
train_df, val_df = train_test_split(
    full_train, test_size=0.2, random_state=42
)

predictor = TabularPredictor(
    label=label,
    problem_type='regression',
    eval_metric='mae'  # ключевая метрика
)

predictor.fit(
    train_data=train_df,
    presets='best_quality',
    num_bag_folds=5,
    num_stack_levels=1)

# 4) Оценка на валидации/тесте
leaderboard = predictor.leaderboard(val_df, silent=True)
print(leaderboard)

performance = predictor.evaluate(val_df)  # выведет MAE (и др. метрики)
print("MAE:", performance["mae"])

# 5) Предсказания
y_pred_val = predictor.predict(val_df.drop(columns=[label]))
print(y_pred_val.head())

# 6) Важность признаков (полезно для интерпретации)
fi = predictor.feature_importance(val_df)
print(fi.head(20))
```
